```python
# importing required libraries
import pandas as pd
from pathlib import Path
```

```python
# 1. Resolve paths cleanly using pathlib
parent_dir = Path().resolve().parent
# Using modern pathlib slash (/) operator to join paths cleanly
data_path = parent_dir / 'data'

print(data_path)
```

```python
# 2. Map all 9 files
files_config = ["olist_customers_dataset.csv", "olist_geolocation_dataset.csv", "olist_order_items_dataset.csv",
                "olist_order_payments_dataset.csv", "olist_order_reviews_dataset.csv", "olist_orders_dataset.csv",
                "olist_products_dataset.csv", "olist_sellers_dataset.csv", "product_category_name_translation.csv"
                ]
for file in files_config:
    # Construct the absolute system file path
    file_path = data_path / file

    if file_path.exists():
        # Read the target CSV file into memory and load it as a structured pandas DataFrame object (df)
        df = pd.read_csv(file_path)

        print("*="*50)
        print(file)
        print("*="*50)

        # Shape of df (rows, cols)
        print("\nShape of df ")
        display(df.shape)

        print("\nConcise summary about df")
        df.info()

        # Duplicate records count
        print("\nDuplicates records found: " , df.duplicated().sum())

        # Null values count
        print("\nMissing Values % in each column")
        print((df.isnull().mean()*100).round(2))

    else:
        print(f"{file} File not found at {file_path} ")
```

```python
# To show duplicate records
for file in files_config:
    file_path = data_path / file
    if file_path.exists():
        # read file into dataframe
        df = pd.read_csv(file_path)
        
        # Duplicate  records count
        duplicate_records = df.duplicated().sum()
        if duplicate_records>0:
            print(f"\nTotal {duplicate_records} duplicate records found in '{file}'")
            display(df[df.duplicated(keep=False)].sort_values(by='geolocation_zip_code_prefix'))
        else:
            print(f"\nNo Duplicate records found in '{file}'")
```
